```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

FAQ_PATH = "/content/LJ_University_Campus_Chatbot/faqs.csv"

faqs = pd.read_csv(FAQ_PATH).fillna("")

# Clean question text and prepare it for matching
faqs["question"] = faqs["question"].astype(str).str.strip()
faqs["processed_question"] = faqs["question"].apply(preprocess_text)

# Remove duplicate questions
faqs = faqs.drop_duplicates(subset=["question"], keep="first")
faqs = faqs.reset_index(drop=True)

vectorizer = TfidfVectorizer(ngram_range=(1, 2))
faq_matrix = vectorizer.fit_transform(faqs["processed_question"])

print("Total FAQs:", len(faqs))
print("TF-IDF matrix shape:", faq_matrix.shape)
print("FAQ departments:", sorted(faqs["department"].unique()))
print("\nSample processed questions:")
print(faqs[["question", "processed_question"]].head(10).to_string(index=False))

assert len(faqs) >= 60, "ERROR: FAQ bank has fewer than 60 questions."
assert faq_matrix.shape[0] == len(faqs), "ERROR: TF-IDF matrix size mismatch."

print("\nSTEP 7A PASSED!")
```

```python
def detect_intent(question):
    text = normalize_language(question).lower()

    if any(word in text for word in [
        "fee", "fees", "cost", "tuition", "how much",
        "ketli", "ketla", "kitni", "kitna"
    ]):
        return "fees"

    if any(phrase in text for phrase in [
        "what courses", "which courses", "available courses",
        "courses available", "course list", "program list"
    ]):
        return "course_list"

    if any(word in text for word in [
        "career", "job", "jobs", "internship", "naukri",
        "after", "pachi", "baad"
    ]):
        return "career"

    if any(word in text for word in [
        "admission", "apply", "application", "eligible", "eligibility"
    ]):
        return "admission"

    if any(word in text for word in [
        "course", "courses", "program", "programs", "degree"
    ]):
        return "course"

    if any(word in text for word in [
        "hostel", "accommodation", "stay"
    ]):
        return "hostel"

    return None


def detect_department(question):
    text = normalize_language(question).lower()

    if any(term in text for term in ["architecture", "barch", "b arch"]):
        return "Architecture"

    if any(term in text for term in ["pharmacy", "bpharm", "b pharm"]):
        return "Pharmacy"

    if any(term in text for term in ["engineering", "engineer"]):
        return "Engineering"

    if any(term in text for term in ["management", "mba"]):
        return "Management"

    if any(term in text for term in ["mca", "bca", "bsc it", "msc it"]):
        return "MCA"

    return None


print("Architecture question:", detect_intent(
    "What courses are available in architecture?"
))
print("General admission question:", detect_department(
    "How do I apply for admission?"
))
print("MCA fees question:", detect_department(
    "MCA ni fees ketli che?"
))
```
