```python
# Detect negative eligibility questions
def is_negative_eligibility_question(question):
    text = normalize_language(question).lower()

    negative_words = ["not", "no", "never", "nahi", "nathi", "na"]

    eligibility_words = [
        "eligible", "eligibility", "admission", "qualify"
    ]

    has_negative = any(word in text.split() for word in negative_words)
    has_eligibility = any(word in text for word in eligibility_words)

    return has_negative and has_eligibility


# Test negative-question detection
negative_tests = [
    "I am not eligible for MCA",
    "I am eligible for MCA",
    "MCA admission requirements"
]

for question in negative_tests:
    print(question, "->", is_negative_eligibility_question(question))
```

```python
# STEP 9A: Initialize Conversation Memory

conversation_memory = {
    "last_department": None,
    "last_intent": None,
    "last_question": None,
    "last_answer": None
}

print("Conversation memory initialized!")

print("\nMemory structure:")
for key, value in conversation_memory.items():
    print(key, ":", value)

print("\nSTEP 9A PASSED!")
```

```python
# STEP 9B: Updated Memory-Aware Chatbot

def chatbot_with_memory(user_question):

    global conversation_memory

    question = user_question.strip()
    lower_question = question.lower()

    # 1. Detect department and intent
    detected_department = detect_department(question)
    detected_intent = detect_intent(question)

    # 2. Identify follow-up questions
    follow_up_words = [
        "and",
        "what about",
        "how about",
        "also",
        "its",
        "their",
        "that",
        "what else",
        "tell me more",
        "and for"
    ]

    is_follow_up = any(
        phrase in lower_question for phrase in follow_up_words
    )

    # 3. Select department
    # Priority: New department > Previous department
    if detected_department:
        selected_department = detected_department
    else:
        selected_department = conversation_memory["last_department"]

    # 4. Expand follow-up question with remembered department
    expanded_question = question

    if is_follow_up and selected_department:
        expanded_question = (
            selected_department + " " + question
        )

    # 5. Find the best FAQ answer
    result = find_best_faq(
        expanded_question,
        department=selected_department
    )

    # 6. Update conversation memory
    if result["department"] not in [None, "General"]:
        conversation_memory["last_department"] = result["department"]

    if result["intent"] is not None:
        conversation_memory["last_intent"] = result["intent"]

    conversation_memory["last_question"] = question
    conversation_memory["last_answer"] = result["answer"]

    return result


print("UPDATED STEP 9B PASSED!")
```

```python
# STEP 9B: Interactive Chat Interface

print("=" * 50)
print("       LJ UNIVERSITY CHATBOT")
print("=" * 50)
print("Type 'exit' to end the conversation.")
print("")

while True:

    user_question = input("You: ")

    if user_question.lower().strip() == "exit":
        print("Bot: Thank you for chatting with LJ University!")
        break

    result = chatbot_with_memory(user_question)

    print("\nBot:", result["answer"])
    print("Confidence:", round(result["confidence"], 3))
    print("-" * 50)
```
